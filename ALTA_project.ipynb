```python
import os
import gc
import json
from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.metrics import f1_score
from sklearn.model_selection import StratifiedKFold
import torch
from torch.utils.data import DataLoader
from transformers import AutoTokenizer

from dataset import ALTAMultiTaskDataset
from model import DialectAwareMultiTaskDeBERTa

# Use the SAME project-level components that produced the trained checkpoints.
# This script does NOT import v10/v11 or any previous inference script.
import main as project_main

# ============================================================
# SELF-CONTAINED CALIBRATION / FINAL TEST INFERENCE v12
#
# Inputs already available:
#   - best_fold_1.pt
#   - best_fold_2.pt
#   - best_fold_3.pt
#   - fold_1_cross_adapt.pt
#   - fold_2_cross_adapt.pt
#   - train.csv
#   - valid.csv (757 labeled rows)
#   - real competition test CSV (1,573 rows)
#
# No training. No new cross-adaptation training. No OOF.
# ============================================================

MODEL_NAME = project_main.MODEL_NAME
MAX_LENGTH = project_main.MAX_LENGTH
BATCH_SIZE = project_main.BATCH_SIZE
DEVICE = project_main.DEVICE
SEED = project_main.SEED
PREPROCESS_TEXT = project_main.preprocess_text
CHECKPOINT_DIR = getattr(project_main, "ALTA_CHECKPOINT_DIR", "checkpoints/task_specific")
TRAIN_PATH = getattr(project_main, "train_path", "train.csv")
VALID_PATH = "valid.csv"
GENERAL_ADAPTER_CHECKPOINT = getattr(
    project_main, "GENERAL_ADAPTER_CHECKPOINT", "checkpoints/"
)
USE_ISARCASM_PRETRAINING = getattr(
    project_main, "USE_ISARCASM_PRETRAINING", True
)

NUM_FOLDS = 3
EXPECTED_VALID_ROWS = 757
EXPECTED_TEST_ROWS = 1573

# Existing cross-adapted caches. We DO NOT train them here.
CACHE_DIRS = (
    "cross_dialect_cache_v6",
    "cross_dialect_cache_v4",
    "cross_dialect_cache_v7",
)

# Coarse but useful inference-time search.
MIX_GRID = (0.0, 0.25, 0.50, 0.75, 1.00)
THRESHOLD_GRID = np.round(np.arange(0.30, 0.701, 0.01), 2)
FINAL_MIN_GAIN = 0.003

OUT_CURRENT = "answer_deberta_current_2adapted_1native.csv"
OUT_OPTIMIZED = "answer_deberta_calibrated_optimized.csv"
OUT_THRESHOLD_ONLY = "answer_deberta_calibrated_current_thresholds.csv"
OUT_BLEND = "answer_deberta_calibrated_blend.csv"
OUT_FINAL = "answer_deberta_calibrated_final.csv"
OUT_REPORT = "answer_deberta_calibration_v12_report.json"

SENTIMENT_ADAPTER_NAMES = ("sentiment_au", "sentiment_uk")
SARCASM_ADAPTER_NAMES = ("sarcasm_au", "sarcasm_uk")
ALL_ADAPTERS = SENTIMENT_ADAPTER_NAMES + SARCASM_ADAPTER_NAMES


def set_seed(seed=42):
    if hasattr(project_main, "set_seed"):
        project_main.set_seed(seed)
    else:
        np.random.seed(seed)
        torch.manual_seed(seed)
        if torch.cuda.is_available():
            torch.cuda.manual_seed_all(seed)


def canonical_variety(value):
    s = str(value).strip().upper()
    if "AU" in s and "UK" not in s:
        return "AU"
    if "UK" in s and "AU" not in s:
        return "UK"
    raise ValueError(f"Unknown variety: {value}")


def checkpoint_path(fold_index):
    return os.path.join(CHECKPOINT_DIR, f"best_fold_{fold_index + 1}.pt")


def require_columns(df, columns, name):
    missing = [c for c in columns if c not in df.columns]
    if missing:
        raise KeyError(f"{name} missing columns: {missing}")


def _find_actual_test_path():
    """Find the real 1,573-row competition test CSV; never use valid.csv."""
    explicit = [
        getattr(project_main, "test_path", None),
        "test.csv",
        "test_data.csv",
        "test_public.csv",
        "official_test.csv",
    ]

    seen = set()

    for raw in explicit:
        if not raw:
            continue
        p = Path(str(raw))
        candidates = [p]
        if not p.is_absolute():
            candidates.extend([Path("/content") / p, Path("/content/drive/MyDrive") / p])

        for cand in candidates:
            try:
                cand = cand.resolve()
            except Exception:
                continue
            key = str(cand)
            if key in seen or not cand.is_file() or cand.name.lower() == "valid.csv":
                continue
            seen.add(key)
            try:
                head = pd.read_csv(cand, nrows=3)
                if not {"source", "variety", "text"}.issubset(head.columns):
                    continue
                full = pd.read_csv(cand)
                if len(full) == EXPECTED_TEST_ROWS:
                    return str(cand), full
            except Exception:
                continue

    roots = [Path("/content")]
    drive = Path("/content/drive/MyDrive")
    if drive.is_dir():
        roots.append(drive)

    matches = []
    for root in roots:
        try:
            files = root.rglob("*.csv")
        except Exception:
            continue
        for cand in files:
            try:
                cand = cand.resolve()
            except Exception:
                continue
            key = str(cand)
            if key in seen or cand.name.lower() in {"train.csv", "valid.csv"}:
                continue
            seen.add(key)
            try:
                head = pd.read_csv(cand, nrows=3)
                if not {"source", "variety", "text"}.issubset(head.columns):
                    continue
                full = pd.read_csv(cand)
                if len(full) == EXPECTED_TEST_ROWS:
                    matches.append((cand, full))
            except Exception:
                continue

    if not matches:
        raise FileNotFoundError(
            "Could not find the real 1,573-row test CSV. "
            "Put the competition test CSV in /content (preferably as test.csv)."
        )

    matches.sort(
        key=lambda item: (
            0 if any(t in item[0].name.lower() for t in ("test", "official")) else 1,
            len(str(item[0])),
        )
    )
    return str(matches[0][0]), matches[0][1]


def load_data():
    train_df = pd.read_csv(TRAIN_PATH)
    valid_df = pd.read_csv(VALID_PATH)

    require_columns(train_df, ["text", "variety", "sentiment", "sarcasm"], "train.csv")
    require_columns(valid_df, ["text", "variety", "sentiment", "sarcasm"], "valid.csv")

    train_df = train_df.dropna(
        subset=["text", "variety", "sentiment", "sarcasm"]
    ).reset_index(drop=True)
    valid_df = valid_df.dropna(
        subset=["text", "variety", "sentiment", "sarcasm"]
    ).reset_index(drop=True)

    for df, name in ((train_df, "train"), (valid_df, "valid")):
        df["sentiment"] = pd.to_numeric(df["sentiment"], errors="raise").astype(int)
        df["sarcasm"] = pd.to_numeric(df["sarcasm"], errors="raise").astype(int)
        if not df["sentiment"].isin([0, 1]).all():
            raise ValueError(f"{name}.csv sentiment must contain only 0/1")
        if not df["sarcasm"].isin([0, 1]).all():
            raise ValueError(f"{name}.csv sarcasm must contain only 0/1")

    test_path, test_df = _find_actual_test_path()
    require_columns(test_df, ["source", "variety", "text"], test_path)
    test_df = test_df.dropna(subset=["text", "variety"]).reset_index(drop=True)
    if len(test_df) != EXPECTED_TEST_ROWS:
        raise RuntimeError(
            f"Expected {EXPECTED_TEST_ROWS} test rows, found {len(test_df)} in {test_path}"
        )

    return train_df, valid_df, test_df, test_path


def make_model():
    return DialectAwareMultiTaskDeBERTa(
        model_name=MODEL_NAME,
        lora_r=8,
        lora_alpha=16,
        lora_dropout=0.05,
        dropout=0.1,
    ).to(DEVICE)


def safe_load_checkpoint(model, path, label):
    ckpt = torch.load(path, map_location=DEVICE, weights_only=False)
    if not isinstance(ckpt, dict) or "model_state_dict" not in ckpt:
        raise KeyError(f"Invalid checkpoint: {path}")

    missing, unexpected = model.load_state_dict(
        ckpt["model_state_dict"], strict=False
    )
    lora_missing = [k for k in missing if "lora_" in k.lower()]
    lora_unexpected = [k for k in unexpected if "lora_" in k.lower()]
    head_missing = [
        k for k in missing
        if "sentiment_head" in k or "sarcasm_head" in k
    ]
    head_unexpected = [
        k for k in unexpected
        if "sentiment_head" in k or "sarcasm_head" in k
    ]

    print(
        f"[{label}] missing={len(missing)} unexpected={len(unexpected)} "
        f"LoRA_missing={len(lora_missing)} LoRA_unexpected={len(lora_unexpected)} "
        f"head_missing={len(head_missing)} head_unexpected={len(head_unexpected)}"
    )

    if lora_missing or lora_unexpected or head_missing or head_unexpected:
        raise RuntimeError(f"{label}: checkpoint mismatch")

    model.eval()
    for p in model.parameters():
        p.requires_grad_(False)
    return ckpt


def make_loader(df, tokenizer):
    x = df.copy()
    x["text"] = x["text"].fillna("").astype(str).apply(PREPROCESS_TEXT)
    ds = ALTAMultiTaskDataset(x, tokenizer, MAX_LENGTH)
    return DataLoader(
        ds,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
    )


def predict_probs(model, loader, route="native"):
    sent_parts = []
    sarc_parts = []

    model.eval()
    with torch.inference_mode():
        for batch in loader:
            input_ids = batch["input_ids"].to(DEVICE, non_blocking=True)
            attention_mask = batch["attention_mask"].to(DEVICE, non_blocking=True)
            variety_ids = batch["variety_ids"].to(DEVICE, non_blocking=True)

            if route == "native":
                routed = variety_ids
            elif route == "cross":
                routed = 1 - variety_ids
            else:
                raise ValueError(route)

            out = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                variety_ids=routed,
            )
            sent_parts.append(
                torch.softmax(out["sentiment_logits"].float(), dim=1)[:, 1]
                .cpu().numpy()
            )
            sarc_parts.append(
                torch.softmax(out["sarcasm_logits"].float(), dim=1)[:, 1]
                .cpu().numpy()
            )

    if not sent_parts:
        raise RuntimeError("No predictions produced")
    return np.concatenate(sent_parts), np.concatenate(sarc_parts)


def ensemble(parts):
    return np.mean(np.stack(parts, axis=0), axis=0)


def get_adapter_keys(state, adapter_name):
    needle = adapter_name.lower()
    keys = [
        k for k in state.keys()
        if "lora_" in k.lower() and needle in k.lower()
    ]
    if not keys:
        raise RuntimeError(f"No LoRA keys found for adapter {adapter_name}")
    return keys


def find_existing_cache(fold_index):
    n = fold_index + 1
    filename = f"fold_{n}_cross_adapt.pt"
    for directory in CACHE_DIRS:
        path = os.path.join(directory, filename)
        if os.path.isfile(path):
            return path
    return None


def load_adapted_cache(fold_index):
    path = find_existing_cache(fold_index)
    if path is None:
        return None
    cache = torch.load(path, map_location="cpu", weights_only=False)
    if "adapted_lora" not in cache:
        raise KeyError(f"Invalid cross-adapt cache: {path}")
    print(f"✓ USING EXISTING CROSS-ADAPT CACHE — FOLD {fold_index + 1}: {path}")
    return cache["adapted_lora"]


def merge_task_mix(original_state, adapted_lora, sentiment_mix, sarcasm_mix):
    merged = {k: v.detach().cpu().clone() for k, v in original_state.items()}

    for adapter in SENTIMENT_ADAPTER_NAMES:
        for k in get_adapter_keys(merged, adapter):
            if k not in adapted_lora:
                raise KeyError(f"Missing adapted sentiment key: {k}")
            w = float(sentiment_mix)
            merged[k] = (
                original_state[k].float() * (1.0 - w)
                + adapted_lora[k].float() * w
            ).to(dtype=original_state[k].dtype)

    for adapter in SARCASM_ADAPTER_NAMES:
        for k in get_adapter_keys(merged, adapter):
            if k not in adapted_lora:
                raise KeyError(f"Missing adapted sarcasm key: {k}")
            w = float(sarcasm_mix)
            merged[k] = (
                original_state[k].float() * (1.0 - w)
                + adapted_lora[k].float() * w
            ).to(dtype=original_state[k].dtype)

    return merged


def official_score(df, sent_probs, sarc_probs, sent_thresholds=None, sarc_thresholds=None):
    sent_probs = np.asarray(sent_probs)
    sarc_probs = np.asarray(sarc_probs)
    if len(df) != len(sent_probs) or len(df) != len(sarc_probs):
        raise ValueError(
            f"Length mismatch: df={len(df)} sent={len(sent_probs)} sarc={len(sarc_probs)}"
        )

    varieties = df["variety"].map(canonical_variety).to_numpy()
    y_sent = df["sentiment"].astype(int).to_numpy()
    y_sarc = df["sarcasm"].astype(int).to_numpy()

    if sent_thresholds is None:
        sent_thresholds = {"AU": 0.50, "UK": 0.50}
    if sarc_thresholds is None:
        sarc_thresholds = {"AU": 0.50, "UK": 0.50}

    sent_pred = np.zeros(len(df), dtype=int)
    sarc_pred = np.zeros(len(df), dtype=int)
    sent_scores = {}
    sarc_scores = {}

    for dialect in ("AU", "UK"):
        mask = varieties == dialect
        sent_pred[mask] = (
            sent_probs[mask] >= float(sent_thresholds[dialect])
        ).astype(int)
        sarc_pred[mask] = (
            sarc_probs[mask] >= float(sarc_thresholds[dialect])
        ).astype(int)

        sent_scores[dialect] = f1_score(
            y_sent[mask], sent_pred[mask], average="macro", zero_division=0
        )
        sarc_scores[dialect] = f1_score(
            y_sarc[mask], sarc_pred[mask], average="macro", zero_division=0
        )

    sent_min = min(sent_scores.values())
    sarc_min = min(sarc_scores.values())
    score = (sent_min + sarc_min) / 2.0

    details = {
        "sentiment_AU": float(sent_scores["AU"]),
        "sentiment_UK": float(sent_scores["UK"]),
        "sarcasm_AU": float(sarc_scores["AU"]),
        "sarcasm_UK": float(sarc_scores["UK"]),
        "sentiment_min": float(sent_min),
        "sarcasm_min": float(sarc_min),
        "official_score": float(score),
    }
    return float(score), details


def optimize_thresholds_task(df, probs, task):
    probs = np.asarray(probs)
    varieties = df["variety"].map(canonical_variety).to_numpy()
    labels = df[task].astype(int).to_numpy()

    best = None
    for au_t in THRESHOLD_GRID:
        m_au = varieties == "AU"
        au_score = f1_score(
            labels[m_au],
            (probs[m_au] >= float(au_t)).astype(int),
            average="macro",
            zero_division=0,
        )
        for uk_t in THRESHOLD_GRID:
            m_uk = varieties == "UK"
            uk_score = f1_score(
                labels[m_uk],
                (probs[m_uk] >= float(uk_t)).astype(int),
                average="macro",
                zero_division=0,
            )
            worst = min(au_score, uk_score)
            # Tie-break toward the standard 0.50 threshold.
            distance = abs(float(au_t) - 0.50) + abs(float(uk_t) - 0.50)
            key = (worst, -distance)
            if best is None or key > best[0]:
                best = (
                    key,
                    {"AU": float(au_t), "UK": float(uk_t)},
                    {"AU": float(au_score), "UK": float(uk_score)},
                )

    return best[1], best[2]


def current_winner_score(valid_df, predictions_by_fold):
    # Existing known winner:
    #   fold 1 = adapted 0.50
    #   fold 2 = adapted 0.50
    #   fold 3 = native
    sent = ensemble([
        predictions_by_fold[0]["adapted_sent"][0.50],
        predictions_by_fold[1]["adapted_sent"][0.50],
        predictions_by_fold[2]["native_sent"],
    ])
    sarc = ensemble([
        predictions_by_fold[0]["adapted_sarc"][0.50],
        predictions_by_fold[1]["adapted_sarc"][0.50],
        predictions_by_fold[2]["native_sarc"],
    ])
    score, details = official_score(valid_df, sent, sarc)
    return sent, sarc, score, details


def collect_validation_predictions(valid_df, tokenizer):
    """Inference-only validation predictions for mix calibration."""
    loader = make_loader(valid_df, tokenizer)
    all_folds = []

    for fold_index in range(NUM_FOLDS):
        path = checkpoint_path(fold_index)
        model = make_model()
        ckpt = safe_load_checkpoint(
            model,
            path,
            f"Fold {fold_index + 1} validation calibration",
        )
        original_state = {
            k: v.detach().cpu().clone()
            for k, v in ckpt["model_state_dict"].items()
        }
        adapted_lora = load_adapted_cache(fold_index)

        fold_result = {
            "native_sent": None,
            "native_sarc": None,
            "adapted_sent": {},
            "adapted_sarc": {},
        }

        # Native baseline.
        p_s, p_r = predict_probs(model, loader, route="native")
        fold_result["native_sent"] = p_s.copy()
        fold_result["native_sarc"] = p_r.copy()

        if adapted_lora is not None:
            # Folds 1/2: evaluate the five cheap mix values.
            for mix in MIX_GRID:
                mix = float(mix)
                if mix == 0.0:
                    fold_result["adapted_sent"][mix] = p_s.copy()
                    fold_result["adapted_sarc"][mix] = p_r.copy()
                    continue

                mixed_state = merge_task_mix(
                    original_state,
                    adapted_lora,
                    sentiment_mix=mix,
                    sarcasm_mix=mix,
                )
                model.load_state_dict(mixed_state, strict=False)
                q_s, q_r = predict_probs(model, loader, route="native")
                fold_result["adapted_sent"][mix] = q_s.copy()
                fold_result["adapted_sarc"][mix] = q_r.copy()

            print(
                f"Fold {fold_index + 1}: collected adaptation mixes {list(MIX_GRID)}"
            )
        else:
            # Fold 3: it remains native; populate all mix keys with native only
            # so the search code remains simple.
            for mix in MIX_GRID:
                mix = float(mix)
                fold_result["adapted_sent"][mix] = p_s.copy()
                fold_result["adapted_sarc"][mix] = p_r.copy()
            print(f"Fold {fold_index + 1}: no cross-adapt cache; native only")

        all_folds.append(fold_result)
        del model, ckpt, original_state, adapted_lora
        gc.collect()
        torch.cuda.empty_cache()

    return all_folds


def search_task_mix(valid_df, predictions_by_fold, task):
    """Search fold1/fold2 adapted mix separately for one task."""
    best = None

    y = valid_df[task].astype(int).to_numpy()
    varieties = valid_df["variety"].map(canonical_variety).to_numpy()

    for mix1 in MIX_GRID:
        for mix2 in MIX_GRID:
            mix1 = float(mix1)
            mix2 = float(mix2)
            if task == "sentiment":
                parts = [
                    predictions_by_fold[0]["adapted_sent"][mix1],
                    predictions_by_fold[1]["adapted_sent"][mix2],
                    predictions_by_fold[2]["native_sent"],
                ]
            else:
                parts = [
                    predictions_by_fold[0]["adapted_sarc"][mix1],
                    predictions_by_fold[1]["adapted_sarc"][mix2],
                    predictions_by_fold[2]["native_sarc"],
                ]

            probs = ensemble(parts)
            au = varieties == "AU"
            uk = varieties == "UK"
            f1_au = f1_score(
                y[au], (probs[au] >= 0.50).astype(int),
                average="macro", zero_division=0,
            )
            f1_uk = f1_score(
                y[uk], (probs[uk] >= 0.50).astype(int),
                average="macro", zero_division=0,
            )
            worst = min(f1_au, f1_uk)
            distance = abs(mix1 - 0.50) + abs(mix2 - 0.50)
            key = (worst, -distance)

            if best is None or key > best[0]:
                best = (
                    key,
                    {"fold1_mix": mix1, "fold2_mix": mix2},
                    probs.copy(),
                    {"AU": float(f1_au), "UK": float(f1_uk)},
                )

    return best[1], best[2], best[3]


def build_test_candidate(test_df, tokenizer, sent_mix_f1, sent_mix_f2, sarc_mix_f1, sarc_mix_f2):
    test_loader = make_loader(test_df, tokenizer)
    sent_parts = []
    sarc_parts = []

    selected_mix_info = []

    for fold_index in range(NUM_FOLDS):
        path = checkpoint_path(fold_index)
        model = make_model()
        ckpt = safe_load_checkpoint(
            model,
            path,
            f"Fold {fold_index + 1} TEST final",
        )
        original_state = {
            k: v.detach().cpu().clone()
            for k, v in ckpt["model_state_dict"].items()
        }
        adapted_lora = load_adapted_cache(fold_index)

        if fold_index == 0:
            sent_mix = float(sent_mix_f1)
            sarc_mix = float(sarc_mix_f1)
        elif fold_index == 1:
            sent_mix = float(sent_mix_f2)
            sarc_mix = float(sarc_mix_f2)
        else:
            sent_mix = 0.0
            sarc_mix = 0.0

        if adapted_lora is not None and fold_index < 2 and (sent_mix > 0 or sarc_mix > 0):
            mixed = merge_task_mix(
                original_state,
                adapted_lora,
                sentiment_mix=sent_mix,
                sarcasm_mix=sarc_mix,
            )
            model.load_state_dict(mixed, strict=False)
            selected_mix_info.append({
                "fold": fold_index + 1,
                "sentiment_mix": sent_mix,
                "sarcasm_mix": sarc_mix,
                "mode": "adapted",
            })
        else:
            selected_mix_info.append({
                "fold": fold_index + 1,
                "sentiment_mix": 0.0,
                "sarcasm_mix": 0.0,
                "mode": "native",
            })

        p_s, p_r = predict_probs(model, test_loader, route="native")
        sent_parts.append(p_s)
        sarc_parts.append(p_r)

        del model, ckpt, original_state, adapted_lora
        gc.collect()
        torch.cuda.empty_cache()

    return ensemble(sent_parts), ensemble(sarc_parts), selected_mix_info


def save_with_thresholds(df, sent_probs, sarc_probs, sent_thresholds, sarc_thresholds, path):
    varieties = df["variety"].map(canonical_variety).to_numpy()
    out = df[["source", "variety", "text"]].copy()
    out["sentiment"] = 0
    out["sarcasm"] = 0

    for dialect in ("AU", "UK"):
        mask = varieties == dialect
        out.loc[mask, "sentiment"] = (
            np.asarray(sent_probs)[mask] >= float(sent_thresholds[dialect])
        ).astype(int)
        out.loc[mask, "sarcasm"] = (
            np.asarray(sarc_probs)[mask] >= float(sarc_thresholds[dialect])
        ).astype(int)

    out.to_csv(path, index=False)
    print(f"Saved: {path} | rows={len(out)}")
    return out


def save_fixed_point5(df, sent_probs, sarc_probs, path):
    return save_with_thresholds(
        df,
        sent_probs,
        sarc_probs,
        {"AU": 0.50, "UK": 0.50},
        {"AU": 0.50, "UK": 0.50},
        path,
    )


def main():
    set_seed(SEED)

    if not torch.cuda.is_available():
        raise RuntimeError("CUDA GPU is required.")

    print("=" * 78)
    print("ALTA 2026 — SELF-CONTAINED CALIBRATION v12")
    print("=" * 78)
    print("NO IMPORT OF v10")
    print("NO TRAINING")
    print("NO OOF")
    print("NO NEW CROSS-ADAPTATION TRAINING")
    print("Model:", MODEL_NAME)
    print("GPU:", torch.cuda.get_device_name(0))
    print("=" * 78)

    train_df, valid_df, test_df, test_path = load_data()
    print("Train rows:", len(train_df))
    print("Valid rows:", len(valid_df))
    print("Test rows:", len(test_df))
    print("Test path:", test_path)

    if len(valid_df) != EXPECTED_VALID_ROWS:
        print(
            f"WARNING: expected {EXPECTED_VALID_ROWS} valid rows, found {len(valid_df)}. "
            "Calibration will still run on the available labeled validation rows."
        )

    tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

    # Required trained checkpoints.
    for fold_index in range(NUM_FOLDS):
        path = checkpoint_path(fold_index)
        if not os.path.isfile(path):
            raise FileNotFoundError(f"Missing checkpoint: {path}")

    # --------------------------------------------------------
    # 1) Collect validation predictions from existing models.
    # --------------------------------------------------------
    predictions_by_fold = collect_validation_predictions(valid_df, tokenizer)

    native_sent = ensemble([
        predictions_by_fold[0]["native_sent"],
        predictions_by_fold[1]["native_sent"],
        predictions_by_fold[2]["native_sent"],
    ])
    native_sarc = ensemble([
        predictions_by_fold[0]["native_sarc"],
        predictions_by_fold[1]["native_sarc"],
        predictions_by_fold[2]["native_sarc"],
    ])
    native_score, native_details = official_score(valid_df, native_sent, native_sarc)

    current_sent, current_sarc, current_score, current_details = current_winner_score(
        valid_df, predictions_by_fold
    )

    print("\nValidation baselines")
    print("Native 3-fold @0.50:", f"{native_score:.4f}")
    print("Known winner (2 adapted + 1 native) @0.50:", f"{current_score:.4f}")
    print("Known winner details:", json.dumps(current_details, indent=2))

    # --------------------------------------------------------
    # 2) Optimize adaptation strength separately for sentiment/sarcasm.
    # --------------------------------------------------------
    sent_choice, opt_sent, sent_mix_scores = search_task_mix(
        valid_df, predictions_by_fold, "sentiment"
    )
    sarc_choice, opt_sarc, sarc_mix_scores = search_task_mix(
        valid_df, predictions_by_fold, "sarcasm"
    )

    print("\nBest sentiment adaptation mix:", sent_choice)
    print("Sentiment AU/UK F1:", sent_mix_scores)
    print("Best sarcasm adaptation mix:", sarc_choice)
    print("Sarcasm AU/UK F1:", sarc_mix_scores)

    # Fixed-threshold score of optimized task-wise mixes.
    opt_fixed_score, opt_fixed_details = official_score(
        valid_df, opt_sent, opt_sarc
    )
    print("Optimized mixes @0.50:", f"{opt_fixed_score:.4f}")

    # --------------------------------------------------------
    # 3) Optimize AU/UK thresholds for the known winner and optimized mix.
    # --------------------------------------------------------
    current_sent_thresholds, current_sent_thr_scores = optimize_thresholds_task(
        valid_df, current_sent, "sentiment"
    )
    current_sarc_thresholds, current_sarc_thr_scores = optimize_thresholds_task(
        valid_df, current_sarc, "sarcasm"
    )
    current_thr_score, current_thr_details = official_score(
        valid_df,
        current_sent,
        current_sarc,
        current_sent_thresholds,
        current_sarc_thresholds,
    )

    opt_sent_thresholds, opt_sent_thr_scores = optimize_thresholds_task(
        valid_df, opt_sent, "sentiment"
    )
    opt_sarc_thresholds, opt_sarc_thr_scores = optimize_thresholds_task(
        valid_df, opt_sarc, "sarcasm"
    )
    opt_thr_score, opt_thr_details = official_score(
        valid_df,
        opt_sent,
        opt_sarc,
        opt_sent_thresholds,
        opt_sarc_thresholds,
    )

    print("\nCurrent-winner optimized thresholds:")
    print("  Sentiment:", current_sent_thresholds, current_sent_thr_scores)
    print("  Sarcasm:", current_sarc_thresholds, current_sarc_thr_scores)
    print("  Score:", f"{current_thr_score:.4f}")

    print("\nOptimized-mix thresholds:")
    print("  Sentiment:", opt_sent_thresholds, opt_sent_thr_scores)
    print("  Sarcasm:", opt_sarc_thresholds, opt_sarc_thr_scores)
    print("  Score:", f"{opt_thr_score:.4f}")

    # --------------------------------------------------------
    # 4) Cheap probability blend between known winner and optimized mix.
    # --------------------------------------------------------
    best_blend = None
    for w in (0.0, 0.25, 0.50, 0.75, 1.0):
        w = float(w)
        b_sent = (1.0 - w) * current_sent + w * opt_sent
        b_sarc = (1.0 - w) * current_sarc + w * opt_sarc
        score, details = official_score(
            valid_df,
            b_sent,
            b_sarc,
            {"AU": 0.50, "UK": 0.50},
            {"AU": 0.50, "UK": 0.50},
        )
        key = (score, -abs(w - 0.0))
        if best_blend is None or key > best_blend[0]:
            best_blend = (key, w, b_sent.copy(), b_sarc.copy(), details)

    print(
        "\nBest current/optimized probability blend @0.50:",
        f"w_optimized={best_blend[1]:.2f}",
        f"score={best_blend[0][0]:.4f}",
    )

    # --------------------------------------------------------
    # 5) Decide what recipe to carry to the real test.
    #    Require a meaningful local gain over the known winner @0.50.
    # --------------------------------------------------------
    if opt_thr_score >= current_thr_score + FINAL_MIN_GAIN:
        selected_name = "optimized_mix_with_dialect_thresholds"
        selected_sent_mix_f1 = sent_choice["fold1_mix"]
        selected_sent_mix_f2 = sent_choice["fold2_mix"]
        selected_sarc_mix_f1 = sarc_choice["fold1_mix"]
        selected_sarc_mix_f2 = sarc_choice["fold2_mix"]
        selected_sent_thresholds = opt_sent_thresholds
        selected_sarc_thresholds = opt_sarc_thresholds
        selected_valid_score = opt_thr_score
        selected_sent_valid = opt_sent
        selected_sarc_valid = opt_sarc
    else:
        selected_name = "known_winner_with_dialect_thresholds"
        selected_sent_mix_f1 = 0.50
        selected_sent_mix_f2 = 0.50
        selected_sarc_mix_f1 = 0.50
        selected_sarc_mix_f2 = 0.50
        selected_sent_thresholds = current_sent_thresholds
        selected_sarc_thresholds = current_sarc_thresholds
        selected_valid_score = current_thr_score
        selected_sent_valid = current_sent
        selected_sarc_valid = current_sarc

    print("\nSELECTED RECIPE:", selected_name)
    print("Selected validation score:", f"{selected_valid_score:.4f}")
    print("Current winner threshold-calibrated score:", f"{current_thr_score:.4f}")
    print(
        "Gain:",
        f"{selected_valid_score - current_thr_score:+.4f}",
    )

    # --------------------------------------------------------
    # 6) Final inference on the REAL 1,573-row test set.
    # --------------------------------------------------------
    selected_test_sent, selected_test_sarc, selected_mix_info = build_test_candidate(
        test_df,
        tokenizer,
        selected_sent_mix_f1,
        selected_sent_mix_f2,
        selected_sarc_mix_f1,
        selected_sarc_mix_f2,
    )

    # Also build the known winner exactly.
    current_test_sent, current_test_sarc, current_mix_info = build_test_candidate(
        test_df,
        tokenizer,
        0.50,
        0.50,
        0.50,
        0.50,
    )

    # Save known winner @0.50 for direct comparison.
    save_fixed_point5(
        test_df, current_test_sent, current_test_sarc, OUT_CURRENT
    )

    # Save threshold-calibrated version of the known winner.
    save_with_thresholds(
        test_df,
        current_test_sent,
        current_test_sarc,
        current_sent_thresholds,
        current_sarc_thresholds,
        OUT_THRESHOLD_ONLY,
    )

    # Save selected optimized/known recipe with its chosen dialect thresholds.
    save_with_thresholds(
        test_df,
        selected_test_sent,
        selected_test_sarc,
        selected_sent_thresholds,
        selected_sarc_thresholds,
        OUT_OPTIMIZED,
    )

    # Blend the selected recipe with the known winner using the validation-selected
    # coarse weight, but keep it as an additional candidate, not the forced final.
    blend_weight = float(best_blend[1])
    blend_test_sent = (1.0 - blend_weight) * current_test_sent + blend_weight * selected_test_sent
    blend_test_sarc = (1.0 - blend_weight) * current_test_sarc + blend_weight * selected_test_sarc
    save_fixed_point5(test_df, blend_test_sent, blend_test_sarc, OUT_BLEND)

    # Final follows the selected recipe. This avoids choosing directly from hidden
    # test predictions; only the labeled 757-row validation set selected the recipe.
    save_with_thresholds(
        test_df,
        selected_test_sent,
        selected_test_sarc,
        selected_sent_thresholds,
        selected_sarc_thresholds,
        OUT_FINAL,
    )

    report = {
        "pipeline": "self-contained calibration v12",
        "model_name": MODEL_NAME,
        "valid_rows": int(len(valid_df)),
        "test_rows": int(len(test_df)),
        "test_path": str(test_path),
        "current_strategy": "fold1 adapted 0.50 + fold2 adapted 0.50 + fold3 native",
        "current_validation_fixed_050": float(current_score),
        "current_validation_threshold_calibrated": float(current_thr_score),
        "optimized_mix_validation_fixed_050": float(opt_fixed_score),
        "optimized_mix_validation_threshold_calibrated": float(opt_thr_score),
        "selected_strategy": selected_name,
        "selected_validation_score": float(selected_valid_score),
        "selected_gain_vs_current_threshold_calibrated": float(
            selected_valid_score - current_thr_score
        ),
        "selected_sentiment_mix": {
            "fold1": float(selected_sent_mix_f1),
            "fold2": float(selected_sent_mix_f2),
            "fold3": 0.0,
        },
        "selected_sarcasm_mix": {
            "fold1": float(selected_sarc_mix_f1),
            "fold2": float(selected_sarc_mix_f2),
            "fold3": 0.0,
        },
        "selected_thresholds": {
            "sentiment": selected_sent_thresholds,
            "sarcasm": selected_sarc_thresholds,
        },
        "current_details": current_details,
        "current_threshold_details": current_thr_details,
        "optimized_fixed_details": opt_fixed_details,
        "optimized_threshold_details": opt_thr_details,
        "selected_mix_search": {
            "sentiment": sent_choice,
            "sentiment_scores": sent_mix_scores,
            "sarcasm": sarc_choice,
            "sarcasm_scores": sarc_mix_scores,
        },
        "coarse_blend_weight_optimized": blend_weight,
        "test_selected_mix_info": selected_mix_info,
        "test_current_mix_info": current_mix_info,
        "outputs": {
            "current": OUT_CURRENT,
            "threshold_only": OUT_THRESHOLD_ONLY,
            "optimized": OUT_OPTIMIZED,
            "blend": OUT_BLEND,
            "final": OUT_FINAL,
        },
    }

    with open(OUT_REPORT, "w", encoding="utf-8") as f:
        json.dump(report, f, indent=2)

    print("\n" + "=" * 78)
    print("CALIBRATION + FINAL TEST INFERENCE COMPLETE")
    print("=" * 78)
    print("Final strategy:", selected_name)
    print("Validation score:", f"{selected_valid_score:.4f}")
    print("Test rows:", len(test_df))
    print("FINAL:", OUT_FINAL)
    print("REPORT:", OUT_REPORT)
    print("=" * 78)


if __name__ == "__main__":
    main()
```

```python
import torch
import pandas as pd

from transformers import AutoTokenizer, AutoModel
from peft import LoraConfig, get_peft_model

from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import accuracy_score


# ============================================================
# CONFIG
# ============================================================
MODEL_PATH = "/content/checkpoints/best_general_adapter.pt"
DATA_PATH = "/content/train.csv"

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

MAX_LENGTH = 128
BATCH_SIZE = 16


# ============================================================
# LOAD CHECKPOINT
# ============================================================
checkpoint = torch.load(
    MODEL_PATH,
    map_location="cpu",
    weights_only=False
)

MODEL_NAME = checkpoint["model_name"]

print("Model:", MODEL_NAME)
print("Best epoch:", checkpoint["best_epoch"])
print("Best F1:", checkpoint["best_f1"])


# ============================================================
# LOAD TOKENIZER
# ============================================================
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)


# ============================================================
# BASE DEBERTA MODEL
# ============================================================
base_model = AutoModel.from_pretrained(
    MODEL_NAME
)


# ============================================================
# RECREATE LoRA
# ============================================================
lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    lora_dropout=0.05,
    bias="none",

    target_modules=[
        "query_proj",
        "key_proj",
        "value_proj",
    ],

    modules_to_save=[],
    task_type="FEATURE_EXTRACTION",
)

encoder = get_peft_model(
    base_model,
    lora_config,
    adapter_name="general"
)


# ============================================================
# FULL MODEL
# ============================================================
class SarcasmModel(torch.nn.Module):

    def __init__(self, encoder):
        super().__init__()

        self.encoder = encoder

        # binary sarcasm classifier
        self.binary_head = torch.nn.Sequential(
            torch.nn.Linear(1024, 1024),
            torch.nn.ReLU(),
            torch.nn.Dropout(0.1),
            torch.nn.Linear(1024, 2)
        )

        # auxiliary heads
        self.fine_sarcasm_head = torch.nn.Sequential(
            torch.nn.Linear(1024, 1024),
            torch.nn.ReLU(),
            torch.nn.Dropout(0.1),
            torch.nn.Linear(1024, 1)
        )

        self.contrastive_projection = torch.nn.Sequential(
            torch.nn.Linear(1024, 1024),
            torch.nn.ReLU(),
            torch.nn.Linear(1024, 256)
        )


    def forward(self, input_ids, attention_mask):

      outputs = self.encoder(
          input_ids=input_ids,
          attention_mask=attention_mask
      )

      # CLS representation
      pooled = outputs.last_hidden_state[:, 0, :].float()

      # 2-class sarcasm prediction
      binary_logits = self.binary_head(pooled)

      return binary_logits


# ============================================================
# CREATE MODEL
# ============================================================
model = SarcasmModel(encoder)


# ============================================================
# LOAD YOUR TRAINED WEIGHTS
# ============================================================
state_dict = checkpoint["model_state_dict"]

missing, unexpected = model.load_state_dict(
    state_dict,
    strict=False
)

print("\nMissing keys:", missing)
print("Unexpected keys:", unexpected)


# ============================================================
# DEVICE
# ============================================================
model.to(DEVICE)
model.eval()

print("\nModel loaded successfully on:", DEVICE)


# ============================================================
# DATASET
# ============================================================
df = pd.read_csv(DATA_PATH)

print("\nDataset size:", len(df))
print("\nSarcasm distribution:")
print(df["sarcasm"].value_counts())


class ALTADataset(Dataset):

    def __init__(self, df, tokenizer):
        self.texts = df["text"].astype(str).tolist()
        self.labels = df["sarcasm"].astype(int).tolist()
        self.tokenizer = tokenizer

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):

        encoded = self.tokenizer(
            self.texts[idx],
            max_length=MAX_LENGTH,
            truncation=True,
            padding="max_length",
            return_tensors="pt"
        )

        return {
            "input_ids": encoded["input_ids"].squeeze(0),
            "attention_mask": encoded["attention_mask"].squeeze(0),
            "label": torch.tensor(
                self.labels[idx],
                dtype=torch.long
            )
        }


dataset = ALTADataset(df, tokenizer)

loader = DataLoader(
    dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)


# ============================================================
# PREDICTION
# ============================================================
all_preds = []
all_labels = []

with torch.no_grad():

    for batch in loader:

        input_ids = batch["input_ids"].to(DEVICE)
        attention_mask = batch["attention_mask"].to(DEVICE)
        labels = batch["label"].to(DEVICE)

        logits = model(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        preds = torch.argmax(logits, dim=1)

        all_preds.extend(
            preds.cpu().numpy()
        )

        all_labels.extend(
            labels.cpu().numpy()
        )


# ============================================================
# ACCURACY
# ============================================================
accuracy = accuracy_score(
    all_labels,
    all_preds
)

print("\n========================================")
print(f"Sarcasm Accuracy: {accuracy:.4f}")
print(f"Sarcasm Accuracy: {accuracy * 100:.2f}%")
print("========================================")
```

```python
# ============================================================
# ALTA 2026
# REPLACE SARCASM COLUMN WITH BEST GENERAL ADAPTER
#
# INPUT:
#   Existing 3-fold answer CSV
#
# OUTPUT:
#   /content/combined_answer.csv
#
# Sentiment column:
#   KEPT EXACTLY AS IT IS
#
# Sarcasm column:
#   REPLACED using best_general_adapter.pt
# ============================================================

import os
import gc
import torch
import pandas as pd
import numpy as np

from transformers import AutoTokenizer, AutoModel
from peft import LoraConfig, get_peft_model
from torch.utils.data import Dataset, DataLoader


# ============================================================
# CONFIGURATION
# ============================================================

# ------------------------------------------------------------
# PUT YOUR EXISTING 3-FOLD ANSWER CSV HERE
#
# Example:
# /content/answer.csv
# /content/answer_deberta_fast_final.csv
# ------------------------------------------------------------

INPUT_CSV = "/content/answer_deberta_fast_2adapted_1native.csv"


# Exact pretrained sarcasm model
MODEL_PATH = (
    "/content/checkpoints/best_general_adapter.pt"
)


# Final output
OUTPUT_CSV = (
    "/content/combined_answer.csv"
)


DEVICE = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

MAX_LENGTH = 128
BATCH_SIZE = 16


# ============================================================
# CHECK FILES
# ============================================================

if not os.path.isfile(INPUT_CSV):
    raise FileNotFoundError(
        f"Input CSV not found:\n{INPUT_CSV}"
    )


if not os.path.isfile(MODEL_PATH):
    raise FileNotFoundError(
        f"Sarcasm checkpoint not found:\n{MODEL_PATH}"
    )


# ============================================================
# LOAD INPUT ANSWER CSV
# ============================================================

df = pd.read_csv(INPUT_CSV)

print("=" * 80)
print("ALTA 2026 — REPLACE SARCASM WITH GENERAL ADAPTER")
print("=" * 80)

print(
    "Input CSV:",
    INPUT_CSV
)

print(
    "Rows:",
    len(df)
)

print(
    "Columns:",
    df.columns.tolist()
)


# ============================================================
# CHECK REQUIRED COLUMNS
# ============================================================

required_columns = [
    "text",
    "sentiment",
    "sarcasm",
]

missing = [
    col
    for col in required_columns
    if col not in df.columns
]

if missing:
    raise KeyError(
        f"Input CSV is missing required columns: {missing}"
    )


# Keep original sentiment exactly.
original_sentiment = (
    df["sentiment"]
    .copy()
)


# ============================================================
# LOAD CHECKPOINT
# ============================================================

print("\nLoading general sarcasm checkpoint...")

checkpoint = torch.load(
    MODEL_PATH,
    map_location="cpu",
    weights_only=False
)

MODEL_NAME = checkpoint["model_name"]

print(
    "Model:",
    MODEL_NAME
)

print(
    "Best epoch:",
    checkpoint.get("best_epoch")
)

print(
    "Best F1:",
    checkpoint.get("best_f1")
)


# ============================================================
# TOKENIZER
# ============================================================

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)


# ============================================================
# BASE DEBERTA MODEL
# ============================================================

base_model = AutoModel.from_pretrained(
    MODEL_NAME
)


# ============================================================
# EXACT LoRA CONFIG
# ============================================================

lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    lora_dropout=0.05,
    bias="none",

    target_modules=[
        "query_proj",
        "key_proj",
        "value_proj",
    ],

    modules_to_save=[],
    task_type="FEATURE_EXTRACTION",
)


# ============================================================
# EXACT GENERAL ADAPTER
# ============================================================

encoder = get_peft_model(
    base_model,
    lora_config,
    adapter_name="general"
)


# ============================================================
# EXACT SARCASM MODEL
# ============================================================

class SarcasmModel(
    torch.nn.Module
):

    def __init__(
        self,
        encoder
    ):

        super().__init__()

        self.encoder = encoder

        # ----------------------------------------------------
        # Binary sarcasm classifier
        # ----------------------------------------------------

        self.binary_head = (
            torch.nn.Sequential(
                torch.nn.Linear(
                    1024,
                    1024
                ),

                torch.nn.ReLU(),

                torch.nn.Dropout(
                    0.1
                ),

                torch.nn.Linear(
                    1024,
                    2
                )
            )
        )

        # ----------------------------------------------------
        # Auxiliary head
        # ----------------------------------------------------

        self.fine_sarcasm_head = (
            torch.nn.Sequential(
                torch.nn.Linear(
                    1024,
                    1024
                ),

                torch.nn.ReLU(),

                torch.nn.Dropout(
                    0.1
                ),

                torch.nn.Linear(
                    1024,
                    1
                )
            )
        )

        # ----------------------------------------------------
        # Contrastive projection
        # ----------------------------------------------------

        self.contrastive_projection = (
            torch.nn.Sequential(
                torch.nn.Linear(
                    1024,
                    1024
                ),

                torch.nn.ReLU(),

                torch.nn.Linear(
                    1024,
                    256
                )
            )
        )


    def forward(
        self,
        input_ids,
        attention_mask
    ):

        outputs = self.encoder(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        # ====================================================
        # EXACT CLS POOLING
        # ====================================================

        pooled = (
            outputs.last_hidden_state[
                :,
                0,
                :
            ]
            .float()
        )

        # ====================================================
        # EXACT BINARY PREDICTION
        # ====================================================

        binary_logits = (
            self.binary_head(
                pooled
            )
        )

        return binary_logits


# ============================================================
# CREATE MODEL
# ============================================================

model = SarcasmModel(
    encoder
)


# ============================================================
# LOAD TRAINED WEIGHTS
# ============================================================

state_dict = (
    checkpoint[
        "model_state_dict"
    ]
)

missing, unexpected = (
    model.load_state_dict(
        state_dict,
        strict=False
    )
)

print("\nMissing keys:", missing)
print(
    "Unexpected keys:",
    unexpected
)


# ============================================================
# MOVE TO GPU
# ============================================================

model.to(DEVICE)
model.eval()

print(
    "\nModel loaded successfully on:",
    DEVICE
)


# ============================================================
# DATASET
#
# IMPORTANT:
# No preprocessing.
# No cleaning.
# No dialect routing.
# Exact raw text from the answer CSV.
# ============================================================

class AnswerDataset(
    Dataset
):

    def __init__(
        self,
        dataframe,
        tokenizer
    ):

        self.texts = (
            dataframe["text"]
            .astype(str)
            .tolist()
        )

        self.tokenizer = tokenizer


    def __len__(self):

        return len(
            self.texts
        )


    def __getitem__(
        self,
        idx
    ):

        encoded = (
            self.tokenizer(
                self.texts[idx],

                max_length=MAX_LENGTH,

                truncation=True,

                padding="max_length",

                return_tensors="pt"
            )
        )

        return {
            "input_ids":
                encoded[
                    "input_ids"
                ].squeeze(0),

            "attention_mask":
                encoded[
                    "attention_mask"
                ].squeeze(0),
        }


# ============================================================
# CREATE DATASET
# ============================================================

dataset = AnswerDataset(
    df,
    tokenizer
)


loader = DataLoader(
    dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)


# ============================================================
# PREDICT SARCASM
# ============================================================

sarcasm_predictions = []


print("\nRunning sarcasm inference...")

with torch.no_grad():

    for batch in loader:

        input_ids = (
            batch[
                "input_ids"
            ].to(DEVICE)
        )

        attention_mask = (
            batch[
                "attention_mask"
            ].to(DEVICE)
        )

        logits = model(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        # EXACT prediction rule
        preds = torch.argmax(
            logits,
            dim=1
        )

        sarcasm_predictions.extend(
            preds.cpu().numpy()
        )


# ============================================================
# CONVERT TO NUMPY
# ============================================================

sarcasm_predictions = np.asarray(
    sarcasm_predictions,
    dtype=int
)


# ============================================================
# CHECK PREDICTION COUNT
# ============================================================

if len(sarcasm_predictions) != len(df):

    raise RuntimeError(
        "Sarcasm prediction count does not "
        "match CSV row count.\n"
        f"Rows: {len(df)}\n"
        f"Predictions: {len(sarcasm_predictions)}"
    )


print(
    "\nSarcasm predictions:",
    len(sarcasm_predictions)
)

print(
    "Predicted sarcasm = 1:",
    int(
        sarcasm_predictions.sum()
    )
)

print(
    "Predicted sarcasm percentage:",
    f"{sarcasm_predictions.mean() * 100:.2f}%"
)


# ============================================================
# REPLACE ONLY SARCASM
# ============================================================

df["sarcasm"] = (
    sarcasm_predictions
)


# ============================================================
# VERIFY SENTIMENT WAS NOT CHANGED
# ============================================================

if not df["sentiment"].equals(
    original_sentiment
):

    raise RuntimeError(
        "ERROR: sentiment column changed!"
    )

print(
    "\n✓ Sentiment column remained EXACTLY unchanged."
)


# ============================================================
# SAVE COMBINED ANSWER
# ============================================================

df.to_csv(
    OUTPUT_CSV,
    index=False
)


# ============================================================
# FINAL CHECK
# ============================================================

saved_df = pd.read_csv(
    OUTPUT_CSV
)


if len(saved_df) != len(df):

    raise RuntimeError(
        "Saved CSV row count mismatch."
    )


if not saved_df["sentiment"].equals(
    original_sentiment
):

    raise RuntimeError(
        "Saved CSV sentiment column "
        "does not match original."
    )


print("\n" + "=" * 80)
print("DONE")
print("=" * 80)

print(
    "Original answer:",
    INPUT_CSV
)

print(
    "Combined answer:",
    OUTPUT_CSV
)

print(
    "Rows:",
    len(saved_df)
)

print(
    "Sentiment column: UNCHANGED"
)

print(
    "Sarcasm column: REPLACED BY GENERAL ADAPTER"
)

print(
    "Final columns:",
    saved_df.columns.tolist()
)

print("=" * 80)


# ============================================================
# FREE MEMORY
# ============================================================

del model
del encoder
del base_model
del tokenizer
del dataset
del loader
del checkpoint

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
```

```python
def predict_ensemble(models, dataloader):
    """
    Average the logits from the 3 trained folds.

    Returns:
        sentiment_logits
        sarcasm_logits
    """

    all_sentiment_logits = []
    all_sarcasm_logits = []

    for model in models:
        model.eval()

        fold_sentiment = []
        fold_sarcasm = []

        with torch.no_grad():

            for batch in dataloader:

                input_ids = batch["input_ids"].to(DEVICE)
                attention_mask = batch["attention_mask"].to(DEVICE)
                variety_ids = batch["variety_ids"].to(DEVICE)

                outputs = model(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    variety_ids=variety_ids
                )

                # Your fold models return two logits
                sentiment_logits = outputs["sentiment_logits"]
                sarcasm_logits = outputs["sarcasm_logits"]

                fold_sentiment.append(
                    sentiment_logits.detach().float().cpu()
                )

                fold_sarcasm.append(
                    sarcasm_logits.detach().float().cpu()
                )

        all_sentiment_logits.append(
            torch.cat(fold_sentiment, dim=0)
        )

        all_sarcasm_logits.append(
            torch.cat(fold_sarcasm, dim=0)
        )

    # Average the 3 folds
    sentiment_logits = torch.stack(
        all_sentiment_logits,
        dim=0
    ).mean(dim=0)

    sarcasm_logits = torch.stack(
        all_sarcasm_logits,
        dim=0
    ).mean(dim=0)

    return sentiment_logits, sarcasm_logits
```
